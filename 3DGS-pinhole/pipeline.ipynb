```python
import subprocess
import os
import pandas as pd

#functions
def get_metrics(out, model_path, look):
    lista = []
    for line in out.decode("utf-8").split("\r\n"):
        # Looking for lines like this:
        # [ITER 500] L1 0.07367 PSNR 19.36433 LPIPS 0.08761 SSIM 0.13648 Gaussians 2085 LOSS 0.23164 EFFICIENCY 0.067663 [31/08 15:13:16]
        if look in line:
            iter, metrics = line.split(look)
            iter = iter.split(" ")[-1].replace("]","")
            metrics = metrics.split(" ")
            l1 = metrics[1]
            psnr = metrics[3]
            lpips = metrics[5]
            ssim = metrics[7]
            gaussians = metrics[9]
            loss = metrics[11]
            eff = metrics[13]
            lista.append([model_path, iter, l1, psnr, lpips, ssim, gaussians, loss, eff])
    return lista

#inits
results_file = 'results.csv'
results_df = pd.read_csv(results_file) if os.path.exists(results_file) else pd.DataFrame(columns=['nome','iter','l1','psnr','lpips','ssim','gaussians', 'loss', 'efficiency'])
results_df = results_df.astype({'iter': int, 'l1': float, 'psnr': float, 'lpips': float, 'ssim': float, 'gaussians': int,'loss': float, 'efficiency':float})

look = " Evaluating train: "

basepaths =[ r"D:\Diego\labs\extended-3dgs\data/custom",
             r"D:\Diego\labs\extended-3dgs\data/mipnerf360",
             r"D:\Diego\labs\extended-3dgs\data/nerf",
             r"D:\Diego\labs\extended-3dgs\data/tanksandtemples"]

for basepath in basepaths:
    datasets = [folder for folder in os.listdir(basepath)]
    full_datasets_path = [os.path.join(basepath,folder) for folder in datasets]
    for idx, dataset in enumerate(datasets):
        if dataset in invalid_list:
            print(f"Skipping {dataset} as it is in the invalid_list.")
            continue
        # train from 0.00050 to 0.00001
        for densify_grad_threshold in densify_grad_thresholds:
            model_path = f"out/{dataset}_NDC_{densify_grad_threshold}"
            cmd = f"python train.py -s {full_datasets_path[idx]} --model_path {model_path} --iterations 500 --densify_from_iter 99 --disable_viewer --densify_grad_threshold {densify_grad_threshold} --experiment {dataset.split("_")[0]}"
            if not results_df["nome"].str.contains(model_path).any():
                print(cmd)
                result = subprocess.run(cmd.split(" "), stdout=subprocess.PIPE, stderr=subprocess.PIPE)
                metrics = get_metrics(result.stdout, model_path, look)
                for metric in metrics:
                    results_df.loc[len(results_df)] = metric
                results_df.to_csv(results_file, index=False)

            model_path = f"out/{dataset}_OBJ_{densify_grad_threshold}"
            cmd = f"python train.py -s {full_datasets_path[idx]} --model_path {model_path} --iterations 500 --densify_from_iter 99 --disable_viewer --use_pinhole --densify_grad_threshold {densify_grad_threshold} --experiment {dataset.split("_")[0]}"
            if not results_df["nome"].str.contains(model_path).any():
                print(cmd)
                result = subprocess.run(cmd.split(" "), stdout=subprocess.PIPE, stderr=subprocess.PIPE)
                metrics = get_metrics(result.stdout, model_path, look)
                for metric in metrics:
                    results_df.loc[len(results_df)] = metric
                results_df.to_csv(results_file, index=False)

results_df
```

```python
import pandas as pd

# Load the uploaded CSV file
file_path = "runs_2_mipnerf.csv"
df = pd.read_csv(file_path)

df["Duration"] = df["Duration"].apply(lambda x: x.replace("min","") if "min" in x else float(x.replace("h",""))*60 if "h" in x else float(x.replace("d",""))*60*24 ).astype(float)

new_names = df["Name"].apply(lambda x: x.split("/")[-1])
ours_duration = df.loc[df["Name"].str.contains("OBJ")]['Duration'].mean()
vanilla_duration = df.loc[df["Name"].str.contains("NDC")]['Duration'].mean()

import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# Criar um mapeamento de cores por dataset (ex: stump, room, kitchen...)
datasets = df["Name"].apply(lambda x: x.split("_")[0].split("/")[-1])  # extrai "stump", "room", etc.
unique_datasets = datasets.unique()

# Gerar paleta de cores
colors = dict(zip(unique_datasets, mcolors.TABLEAU_COLORS))

# Plot
plt.figure(figsize=(12,6))
plt.bar(df["Name"].apply(lambda x: x.split("/")[-1]).str.replace("_8_NDC"," 3DGS").str.replace("_8_OBJ_7e-05"," Ours"), df["Duration"], color=[colors[d] for d in datasets])
plt.xticks(rotation=90)
plt.xlabel("Dataset")
plt.ylabel("Duração (min)", rotation=0, labelpad=30)
plt.title("Gaussian Splatting: Por que não o modelo de câmera de orifício? \n\nComparação de tempo de processamento até 30K por dataset do MipNerf360 (Ours max_grad_threshold = 7e-5)")
plt.tight_layout()
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', color='gray', alpha=0.7)

# Horizontal line at mean
plt.axhline(y=vanilla_duration, color='r', linestyle='--', label='Duração média (vanilla 3DGS)')
plt.axhline(y=ours_duration, color='b', linestyle='--', label='Duração média (ours)')

plt.legend()
plt.show()
print("Redução de:", 1-ours_duration/vanilla_duration)
```
